```python
import numpy as np
import galois
```

```python
GF2 = galois.GF(2)
```

```python
l = m = 4
```

```python

p_l = [(i+1)%l for i in range(l)]
p_m = [(i+1)%l for i in range(m)]

I_l = GF2.Identity(l)
I_m = GF2.Identity(m)

S_l = GF2.Identity(l)[p_l]
S_m = GF2.Identity(l)[p_m]

x = np.kron(S_l,I_m)
y = np.kron(I_l,S_m)

perms = [x,y]
```

```python
A_vars = [[0,1], [1,2], [1,3]]
```

```python
A = GF2.Zeros(x.shape)
for mat, p in A_vars:
    A += np.power(perms[mat],p)
```
